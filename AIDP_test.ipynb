```
from pyspark.sql import functions as F

table_name = "default.user_info"

df = (
    spark.range(1, 101)
    .withColumnRenamed("id", "id")
    .withColumn("name", F.concat(F.lit("name_"), F.col("id").cast("string")))
    .withColumn("address", F.concat(F.lit("address_"), F.col("id").cast("string")))
    .select("id", "name", "address")
)

spark.sql(f"DROP TABLE IF EXISTS {table_name}")
df.write.mode("overwrite").saveAsTable(table_name)

spark.table(table_name).show(10, truncate=False)
print(f"Row count in {table_name}:", spark.table(table_name).count())
```
