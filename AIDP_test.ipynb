```
import sys
# 验证 sqlite3 是否已存在于 Python 标准库中
import sqlite3
print(sqlite3.sqlite_version)
```

```


# Create/connect to a local SQLite database file
conn = sqlite3.connect("example.db")
cur = conn.cursor()

# Create table
cur.execute("""
CREATE TABLE IF NOT EXISTS people (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    address TEXT NOT NULL
)
""")

# Insert 100 rows
rows = [(i, f"name_{i:03d}", f"address_{i:03d}") for i in range(1, 101)]
cur.executemany("INSERT OR REPLACE INTO people (id, name, address) VALUES (?, ?, ?)", rows)
conn.commit()

# Verify
cur.execute("SELECT COUNT(*) FROM people")
count = cur.fetchone()[0]
count
```
